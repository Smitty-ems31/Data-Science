```python
# ---------------------------------------------------------------------------
# Question 1 – Load Text
# ---------------------------------------------------------------------------
BaseFilePath = Path(r"D:\GradSchool\DAAN862\Lesson12\Assignment_12.txt")

file = open(BaseFilePath , 'r')

text1 = file.read()

file.close()

print("\n=== Output 500 characters to verifty import ===")
print(text1[:500])
```

```python
# ---------------------------------------------------------------------------
# Question 2 – File-name Pattern & Match Count
# ---------------------------------------------------------------------------

import re


pattern = r'[A-Z0-9._%+-]+\.[A-Z]{2,4}'

# Compile with IGNORECASE, just like the Lesson 12 email example
regex = re.compile(pattern, flags=re.IGNORECASE)

# Find all filenames that match this pattern anywhere in text1
filenames_q2 = regex.findall(text1)

print("\n=== File-name Pattern & Match Count ===")
print("Pattern used for filenames:", regex.pattern)
print("Number of filenames matching the pattern:", len(filenames_q2))
print("First 10 matches:", filenames_q2[:10])
```

```python
# ---------------------------------------------------------------------------
# Question 3 – Filenames That Do NOT Match the Pattern
# ---------------------------------------------------------------------------

# Split the text into candidate "filenames" (likely one per line or separated by spaces)
all_names = text1.split()   # whitespace split is enough for this assignment

non_matching = []

for name in all_names:
    # fullmatch: the whole token must match the pattern to count as a "valid" filename
    if regex.fullmatch(name) is None:
        non_matching.append(name)

print("\n=== Filenames That Do NOT Match the Pattern ===")
print("Total tokens (candidate filenames):", len(all_names))
print("Number of filenames that DO NOT match the pattern:", len(non_matching))
print("First 10 non-matching names:", non_matching[:10])
```

```python
# ---------------------------------------------------------------------------
# Question 4 – Word Normalization & Counts
# ---------------------------------------------------------------------------
import nltk

BaseFilePath2 = Path(r"D:\GradSchool\DAAN862\Lesson12\arxiv_annotate1_13_1.txt")
file = open(BaseFilePath2, 'r')

text2 = file.read()

file.close()

tokens = nltk.word_tokenize(text2)

tokens_lower = [w.lower() for w in tokens]

dist = nltk.FreqDist(tokens_lower)

print("\n=== Word Normalization & Counts ===")
print("Total number of normalized tokens:", len(tokens_lower))
print("Number of unique normalized words:", len(dist))
print("Most common normalized words and their counts:")
print(dist.most_common(20))
```
